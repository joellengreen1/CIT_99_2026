# Week 15 Introduction to Neural Networks
Welcome to the class, everyone! Today, we’re going to talk about one of the most powerful tools in a data scientist's toolkit: **Artificial Neural Networks (ANNs)**.

Even though it sounds like something out of a sci-fi movie, the idea is actually inspired by something you already own—your brain!

---

## 1. The Inspiration: Your Brain
Inside your head, you have billions of tiny cells called **neurons**. When you see a soccer ball flying toward you, these neurons send electrical signals to each other to help you decide what to do: *Should I duck? Should I kick it?*

An **Artificial** Neural Network is basically a computer program that tries to mimic that process to solve problems, like identifying a cat in a photo or predicting the price of a house.

---

## 2. How It’s Built (The "Layers")
Think of an ANN like a team of people standing in lines (we call these **layers**). Information passes from the front of the room to the back.

* **The Input Layer:** This is the "eyes" of the network. If we’re teaching the computer to recognize a handwritten number "5," this layer looks at every tiny pixel of the image.
* **The Hidden Layers:** This is where the "thinking" happens. These middle layers look for patterns. One layer might look for straight lines, the next might look for curves, and the next might see how those lines and curves fit together. 
* **The Output Layer:** This is the final answer. The computer says, "I’m 98% sure this is a 5!"



---

## 3. How It Learns: Weights and Bias
Imagine you’re learning to bake a cake. The first time you try, you might add way too much salt. It tastes terrible! So, you adjust your "formula" and add less salt next time.

In a Neural Network, every connection between those "people" in line has a **Weight**.
* A **high weight** means that piece of information is very important (like the shape of the number).
* A **low weight** means the information isn't very helpful (like the color of the paper).

When the network gets an answer wrong, it goes back and tweaks those weights—just like you tweaking your cake recipe—until it starts getting the answers right.

---

### Why this matters
We use these networks every day. When Netflix suggests a movie you might like, or when your phone unlocks using your face, there is a Neural Network under the hood, calculating weights and patterns just like we practiced!

We’ve already using a **Pipeline** to organize our code. Now, let’s pull back the curtain on that `MLPClassifier` we just used. 

Think of the `MLPClassifier` as a **multi-layered filter system**. "MLP" stands for **Multi-Layer Perceptron**, which is just a fancy, "data scientist" way of saying a "Neural Network with multiple steps."

Here is how it works:

---

## 1. The Input: The "Features"
Imagine we are trying to decide if a cell is healthy or cancerous based on the breast cancer data. 
* We feed the computer "clues" (the features): how big is the cell? Is it bumpy or smooth? Is it perfectly round?
* These go into the **Input Layer**.

## 2. The Hidden Layers: The "Internal Detectives"
This is the "MLP" part. Instead of one person making a guess, we have groups of "detectives" in the middle.
* **Layer 1:** One detective might only look for "is it big AND bumpy?"
* **Layer 2:** Another detective takes the notes from the first group and asks, "If it's big, bumpy, and has rough edges, then I'm getting worried."

By having multiple **Hidden Layers**, the computer can find patterns that are too complex for a human to see just by looking at a spreadsheet.



## 3. The Activation: The "Spark"
Inside each detective's head is something called an **Activation Function** (in our code, we used `relu`). 
* Think of this like a light switch. 
* If the clues aren't strong enough, the switch stays **OFF** (0). 
* If the clues are strong enough, the switch flips **ON** and passes the information to the next layer. 

## 4. The Solver: The "Teacher"
In our code, we used a solver called `adam`. 
* Imagine you’re playing a game of "Hot or Cold" to find a hidden object. 
* Every time the model makes a guess, the **Solver** tells it how far off it was. 
* If the model guessed "Healthy" but the answer was "Cancer," the Solver goes back and says, "Hey, change your weights! Pay more attention to the bumpiness next time!"

---

## Summary
* **Multi-Layer:** It has many rows of "detectives" (neurons).
* **Perceptron:** A single "detective" making a tiny decision.
* **Classifier:** The final goal—putting the data into a bucket (Class 0 or Class 1).

Since we are dealing with a medical scenario—trying to determine if a tumor is malignant (cancerous) or benign (healthy)—we don't just want to know if the model is generally "right." We want to know *how* it is right.

For this model, we typically look at four main metrics:

### 1. Accuracy (The "Overall Grade")
This is the simplest metric. It tells us the percentage of total guesses the model got correct.
* **The Question:** "Out of 100 patients, how many did the model label correctly?"
* **The Catch:** If 95% of your patients are healthy, a model could just guess "Healthy" for everyone and get a 95% accuracy without actually finding a single case of cancer!

### 2. Precision (The "False Alarm" Check)
This looks at everyone the model labeled as "Cancer."
* **The Question:** "Of all the people the model flagged for cancer, how many actually had it?"
* **Why it matters:** High precision means fewer "False Alarms" (telling a healthy person they have cancer).

### 3. Recall (The "Safety" Check)
This is often the **most important** metric for medical data.
* **The Question:** "Of all the people who actually had cancer, how many did the model catch?"
* **Why it matters:** High recall means we didn't "miss" anyone. In medicine, missing a cancer diagnosis (a False Negative) is much more dangerous than a false alarm.

### 4. F1-Score (The "Balance")
The F1-Score is a blend of Precision and Recall. It’s a single number that tells us if the model is doing a good job of balancing both.
